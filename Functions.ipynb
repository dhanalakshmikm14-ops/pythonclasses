instead of writing the same code again we define it once and reuse it.

why fuctions matter for data analysts?

-->reuse logic(cleaning,calculation)

Defining functions:

Functions are defined using the def keyword

syntax:
def function_name():
    code


```python
#simple function example

def greet():
    print("Hello")        #defining a function

greet()           #calling a function
```

```python
#function with parameters

def greet(name):
    print("Hello")  #defining function

greet("sri")
    
```

parameters and aruguments

parameters-
parameters are placeholders written inside the function

example-

def add(a,b):
    return a+b

here a and b are parameters
they don't have real values yet

think of it like these are empty boxes waiting for values

Arguments-
Arguments are the actual values passed to the function.

example-
add(2,3)   #calling a function

here 2 and 3 are arguments
they fill the parameters a and b.

Parameters are used when defining a function
Arguments are used when calling  a function


```python
#Function arguments
#positional vs keyword arguments

#positional arguments
#here order matters

def add(a,b):
    print(a+b)
add(2,3)
```

```python
#keyword arguments 

def add(a,b):
    print(a+b)

add(b=3,a=2)
#here order does not matter

#when to use keyword arguments-
#when many parameters 
#when meaning matters 
#when writing clean,readable code
```

Return values-

what does return do?
sends a vlue back to the caller
ends the function
allows further computaation


```python
#Return values example 1: function with print only

def add(a,b):
    print(a+b)
result=add(2,3)
print(result)
```

```python
#function with return

def add(a,b):
    return a+b
result=add(2,3)
print(result)
```

```python
def square(n):
    return n*n
result=square(4)
print(result)

#function calulates 4*4
#return 16 sends value back
#result=16
```

```python
#code after return never runs

def test():
    print("Before return")
    return 10
    print("after return")  #never runs

x=test()
print(x)

```

```python
#returning multiple values

def calculate(a,b):
    total=a+b
    diff=a-b
    return total,diff

result=calculate(10,5)
print(result)
```

Documentation

writing docstrings

A docstring explains:
-what the funtion does
-parameters
-return value

it helps:

other developers
your future self
documentation tools

Basic docstring syntax:
def function():
    """
    Description
    """

```python
#google-stlye docstring(beginner-friendly)

def add(a,b):
    """
    Adds two numbers.

    Args:
    a(int):First number
    b(int):Second number
    Returns:
    int:sum of a and b
    """

    return a+b
```

```python
#numpy-style docstring(industry common)

def add(a,b):
    """
    Adds two numbers.

    parameters
    ----------
    a : int
        First number
    b : int
        Second number
    Returns
    -------
    int
       sum of a and b
    """"
    return a + b
```

```python
#to view docstring
help(add)
```

```python
#sales summary function

def sales_summary(sales):
    total=0
    highest=sales[0]
    lowest=sales[0]

    for amount in sales:
        total=total+amount

        if amount>highest:
            highest=amount
        if amount<lowest:
            lowest=amount

    average=total/len(sales)
    return total,highest,lowest,average

summary=sales_summary([100,200,150,300])
print(summary)
```

```python
#attendance percentage

def attendance_percentage(attended,total):
    if total==0:
        return 0

    percentage=(attended/total)*100
    
    if percentage>=75:
        status="Eligible"
    else:
        status="Not eligible"

    return percentage,status
p,s=attendance_percentage(18,20)
print(p,s)
```
