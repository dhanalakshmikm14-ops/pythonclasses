Loops

A loop helps us repeat the same task again and again without writing  the  same code many times

Types of loop:
for loop
while loop


for loop: a for loop takes one item at a time from collection  and works on it

syntax:
 for item in collection:
    print(item)

collection-list/string/range
item-temporary variable

```python
#basic examples

for i in range(5):
    print(i)

#range (5) is numbers from 0 to 4
# i takes one value at a time
# print(i) runs     
```

```python
for i in range(1,6):
    print(i)
```

```python
#for loop with list items

fruits=["apple","banana","mango"]
for i in fruits:
    print(i)
```

```python
#For loop with string

name="python"

for letter in name:
    print(letter)

#strings are collection of characters
```

```python
#numbers list

numbers=[10,20,30]

for n in numbers:
    print(n*2)
```

```python
#for loop with a dictionary

student={"name":"alex","age":22}
for key in student:
    print(key)
```

```python
student={"name":"alex","age":22}
for value in student.values():
    print(value)
```

```python
#Looping through key value pairs

for key,value in student.items():
    print(key,":",value)
```

```python
#print even numbers

for i in range(0,11,2):
    print(i)
```

WHILE  LOOP

A while loop keeps running as long as a condition remains true

A for loop is used when you know how mny times to repeat

A while loop is used when you dont know how many times to repeat 

Example :

for loop: "Attend 30 students"
you know the number

while loop:"keep serving customers untill shop closes"
you dont know how many
 
 syntax:

while<expression>:
    statement(s)


```python
#counter controlled loop - when you know how many times the loop must run 

i=0
while(i<=10):
    print(i)
    i=i+1
```

```python
#condition controlled lop when you dont know how many times the loop must run 

i = int(input("enter a number :"))
while(i!=10):
    print(i)
    i=int(input("enter a number:"))
```

```python
#you dont know when it will stop

while True:
    user = input("enter the password")

    if user=="admin":
        break
```

```python
#data validation  

age=int(input("enter age"))

while age<0:
    print("invalid age")
    age=int(input("enter age again:"))
    
```

```python
#why for loop is not ideal for an ended input

for i in range(300):
    name=input("enter name")
    if name=="exit":
        break

#what if user needs more inputs, what if the user needs fewer inputs
#this number would be fake and random and its not a good practice
```

```python
while True:
    name=input("enter name(type exit to stop):")
    if name=="exit":
        break

    print("hello",name)

#no fake limit
#user decides when to stop
#logic matches real requirement
```

Control flow keywords

Break- stop the loop completely
break immediately exist the loop
once break runs ,the loop stops forever

continue-skip the current iteration
continue skips the current loop step and moves to the next one,
the loop doent stop

pass-pass does nothing .It is used when python expects code but you're not ready to write it 
yet.Used like a place holder,
The loop does not stop here also

break-->stop the loop
continue-->skip this step
pass-->do nothing


```python
#break,continue and pass

#break-
for i in range(5):
    if i==3:
        break
    print(i)
    

```

```python
#continue

for i in range(5):
    if i==3:
        continue
    print(i)
```

```python
#pass

for i in range(5):
    if i==3:
        pass
    else:
        print(i)
```

#List comprehensions

[x for x in list if condition]

List comprehension is a shorter way to create a new list using logic

```python
#normal way

numbers=[1,2,3]
squares=[]

for n in numbers:
    squares.append(n*n)
print(squares)
```

```python
numbers=[1,2,3]
squares=[n*n for n in numbers]

print(numbers)
print(squares)
```

```python
#with condition

evens=[n for n in numbers if n%2==0]
evens
```

```python
#dictionary comprehension

numbers=[1,2,3]
result={}

for n in numbers:
    result[n]=n*n
result
```

```python
#comprehension way

result={n:n*n for n in numbers}
result
```

```python
#with condition

even_squares={n:n*n for n in numbers if n%2==0}
even_squares
```

NESTED LOOPS
A LOOP  WITHIN ANOTHER LOOP

Used when data contains data inside it or when rows contain column


```python
#EXAMPLE FOR NESTED LOOP

data=[
    [10,20],
    [30,40]
]

for row in data:
    for value in row:
        print(value)

#outer loop is for rows here
#inner loop is for values inside each row
```

```python
#total of each row 


data=[
    [10,20],
    [30,40]
]

for row in data:
    total=0
    for value in row:
        total+=value
    print(total)
```

```python
#small numbers

for i in range(2):
    for j in range(2):
        print(i,j)
```

```python
#count values greater than 20

data=[
    [10,25],
    [30,5]
]

count=0

for row in data:
    for value in row:
        if value>20:
            count=count+1
print(count)
```

```python
#working with strings
#characters inside words

words=["hi","bye"]
for word in words:
    for letter in word:
        print(letter)
```

```python
#comparing two lists

a=[1,2]
b=[3,4]

for x in a:
    for y in b:
        print(x,y)
```

```python
#finding matching values

a=[1,2,3]
b=[2,3,4]

for i in a:
    for j in b:
        if i==j:
            print("match found",i)
```

```python
#nested loops with dictionaries
#dictionary with lists

students={
    "Alex":[80,90],
    "Sam":[50,60]
}

for name,marks in students.items():
    for value in marks:
        print(name,value)
```

```python
#square pattern

for i in range(4):
    for j in range(4):
        print("*",end="")
    print()
```
